```python
# Scheduled Caste Population Growth & Demographic Forecasting System
# Author: Sagnik Patra

import os
import json
import yaml
import pickle
import warnings
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

warnings.filterwarnings("ignore")

# =========================
# PATHS
# =========================

DATASET_PATH = r"C:\Users\NXTWAVE\Downloads\Scheduled Caste Population Growth & Demographic Forecasting System\Percentage_of_Scheduled_Caste_Population_to_Total_Population.csv"

OUTPUT_DIR = r"C:\Users\NXTWAVE\Downloads\Scheduled Caste Population Growth & Demographic Forecasting System"

os.makedirs(OUTPUT_DIR, exist_ok=True)

PREFIX = "sc_"

# =========================
# LOAD DATA
# =========================

df = pd.read_csv(DATASET_PATH)
df.columns = df.columns.str.strip()

print("Dataset Loaded Successfully")
print(df.head())
print(df.info())

# =========================
# FEATURE ENGINEERING
# =========================

district_col = "District"

numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()

df["Rural_Growth"] = (
    df["Percentage of Scheduled Caste Population _2011_Rural"]
    - df["Percentage of Scheduled Caste Population _2001_Rural"]
)

df["Urban_Growth"] = (
    df["Percentage of Scheduled Caste Population _2011_Urban"]
    - df["Percentage of Scheduled Caste Population _2001_Urban"]
)

df["Total_Growth"] = (
    df["Percentage of Scheduled Caste Population _2011_Total"]
    - df["Percentage of Scheduled Caste Population _2001_Total"]
)

df["Rural_Urban_Gap_2011"] = (
    df["Percentage of Scheduled Caste Population _2011_Rural"]
    - df["Percentage of Scheduled Caste Population _2011_Urban"]
)

# Forecast target: estimated 2021 total SC percentage
df["Estimated_2021_Total_SC_Percentage"] = (
    df["Percentage of Scheduled Caste Population _2011_Total"] + df["Total_Growth"]
)

# =========================
# FEATURES AND TARGET
# =========================

feature_cols = [
    "Percentage of Scheduled Caste Population _2001_Rural",
    "Percentage of Scheduled Caste Population _2001_Urban",
    "Percentage of Scheduled Caste Population _2001_Total",
    "Percentage of Scheduled Caste Population _2011_Rural",
    "Percentage of Scheduled Caste Population _2011_Urban",
    "Percentage of Scheduled Caste Population _2011_Total",
    "Rural_Growth",
    "Urban_Growth",
    "Total_Growth",
    "Rural_Urban_Gap_2011",
]

target_col = "Estimated_2021_Total_SC_Percentage"

X = df[feature_cols]
y = df[target_col]

# =========================
# TRAIN TEST SPLIT
# =========================

X_train, X_test, y_train, y_test, district_train, district_test = train_test_split(
    X,
    y,
    df[district_col],
    test_size=0.2,
    random_state=42
)

# =========================
# SCALING
# =========================

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Save scaler
scaler_path = os.path.join(OUTPUT_DIR, PREFIX + "scaler.pkl")
with open(scaler_path, "wb") as f:
    pickle.dump(scaler, f)

# =========================
# ML MODELS
# =========================

models = {
    "RandomForest": RandomForestRegressor(
        n_estimators=200,
        random_state=42
    ),
    "GradientBoosting": GradientBoostingRegressor(
        n_estimators=200,
        learning_rate=0.05,
        random_state=42
    )
}

results = []

best_model = None
best_model_name = None
best_r2 = -999

for name, model in models.items():
    model.fit(X_train_scaled, y_train)

    preds = model.predict(X_test_scaled)

    mae = mean_absolute_error(y_test, preds)
    mse = mean_squared_error(y_test, preds)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_test, preds)

    results.append({
        "Model": name,
        "MAE": float(mae),
        "MSE": float(mse),
        "RMSE": float(rmse),
        "R2_Score": float(r2)
    })

    model_path = os.path.join(OUTPUT_DIR, PREFIX + name.lower() + "_model.pkl")
    with open(model_path, "wb") as f:
        pickle.dump(model, f)

    if r2 > best_r2:
        best_r2 = r2
        best_model = model
        best_model_name = name

# =========================
# DEEP LEARNING MODEL
# =========================

dl_model = Sequential([
    Dense(64, activation="relu", input_shape=(X_train_scaled.shape[1],)),
    Dropout(0.2),
    Dense(32, activation="relu"),
    Dense(16, activation="relu"),
    Dense(1)
])

dl_model.compile(
    optimizer="adam",
    loss="mse",
    metrics=["mae"]
)

early_stop = EarlyStopping(
    monitor="val_loss",
    patience=20,
    restore_best_weights=True
)

history = dl_model.fit(
    X_train_scaled,
    y_train,
    validation_split=0.2,
    epochs=300,
    batch_size=8,
    callbacks=[early_stop],
    verbose=1
)

dl_preds = dl_model.predict(X_test_scaled).flatten()

dl_mae = mean_absolute_error(y_test, dl_preds)
dl_mse = mean_squared_error(y_test, dl_preds)
dl_rmse = np.sqrt(dl_mse)
dl_r2 = r2_score(y_test, dl_preds)

results.append({
    "Model": "DeepNeuralNetwork",
    "MAE": float(dl_mae),
    "MSE": float(dl_mse),
    "RMSE": float(dl_rmse),
    "R2_Score": float(dl_r2)
})

h5_path = os.path.join(OUTPUT_DIR, PREFIX + "deep_learning_model.h5")
dl_model.save(h5_path)

# =========================
# BEST MODEL SAVE
# =========================

best_model_path = os.path.join(OUTPUT_DIR, PREFIX + "best_model.pkl")
with open(best_model_path, "wb") as f:
    pickle.dump(best_model, f)

# =========================
# PREDICTIONS CSV
# =========================

best_predictions = best_model.predict(X_test_scaled)

prediction_df = pd.DataFrame({
    "District": district_test.values,
    "Actual_2021_Estimated_SC_Percentage": y_test.values,
    "Predicted_2021_SC_Percentage": best_predictions
})

prediction_csv_path = os.path.join(OUTPUT_DIR, PREFIX + "predictions.csv")
prediction_df.to_csv(prediction_csv_path, index=False)

# =========================
# RESULTS CSV
# =========================

results_df = pd.DataFrame(results)

results_csv_path = os.path.join(OUTPUT_DIR, PREFIX + "results.csv")
results_df.to_csv(results_csv_path, index=False)

# =========================
# JSON RESULT FILE
# =========================

json_data = {
    "project_name": "Scheduled Caste Population Growth & Demographic Forecasting System",
    "author": "Sagnik Patra",
    "dataset_path": DATASET_PATH,
    "output_directory": OUTPUT_DIR,
    "target_column": target_col,
    "features_used": feature_cols,
    "best_model": best_model_name,
    "best_r2_score": float(best_r2),
    "model_results": results
}

json_path = os.path.join(OUTPUT_DIR, PREFIX + "results.json")
with open(json_path, "w") as f:
    json.dump(json_data, f, indent=4)

# =========================
# YAML CONFIG FILE
# =========================

yaml_config = {
    "project": {
        "name": "Scheduled Caste Population Growth & Demographic Forecasting System",
        "author": "Sagnik Patra",
        "prefix": PREFIX
    },
    "paths": {
        "dataset_path": DATASET_PATH,
        "output_directory": OUTPUT_DIR,
        "h5_model": h5_path,
        "best_model_pkl": best_model_path,
        "scaler_pkl": scaler_path,
        "results_csv": results_csv_path,
        "predictions_csv": prediction_csv_path,
        "json_results": json_path
    },
    "training": {
        "test_size": 0.2,
        "random_state": 42,
        "target_column": target_col,
        "features": feature_cols
    },
    "models": list(models.keys()) + ["DeepNeuralNetwork"],
    "best_model": best_model_name
}

yaml_path = os.path.join(OUTPUT_DIR, PREFIX + "config.yaml")
with open(yaml_path, "w") as f:
    yaml.dump(yaml_config, f, default_flow_style=False)

# =========================
# FINAL OUTPUT
# =========================

print("\nAll files saved successfully!")
print("Saved Files:")
print(h5_path)
print(best_model_path)
print(scaler_path)
print(results_csv_path)
print(prediction_csv_path)
print(json_path)
print(yaml_path)

print("\nModel Results:")
print(results_df)
```
