```python
# Scheduled Caste Population Growth & Demographic Forecasting System
# Author: Sagnik Patra

import os
import json
import yaml
import pickle
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

warnings.filterwarnings("ignore")

# =========================
# PATHS
# =========================

DATASET_PATH = r"C:\Users\NXTWAVE\Downloads\Scheduled Caste Population Growth & Demographic Forecasting System\Percentage_of_Scheduled_Caste_Population_to_Total_Population.csv"

OUTPUT_DIR = r"C:\Users\NXTWAVE\Downloads\Scheduled Caste Population Growth & Demographic Forecasting System"

os.makedirs(OUTPUT_DIR, exist_ok=True)

PREFIX = "sc_"

# =========================
# LOAD DATA
# =========================

df = pd.read_csv(DATASET_PATH)
df.columns = df.columns.str.strip()

print("Dataset Loaded Successfully")
print(df.head())
print(df.columns)

# =========================
# FEATURE ENGINEERING
# =========================

district_col = "District"

df["Rural_Growth"] = (
    df["Percentage of Scheduled Caste Population _2011_Rural"]
    - df["Percentage of Scheduled Caste Population _2001_Rural"]
)

df["Urban_Growth"] = (
    df["Percentage of Scheduled Caste Population _2011_Urban"]
    - df["Percentage of Scheduled Caste Population _2001_Urban"]
)

df["Total_Growth"] = (
    df["Percentage of Scheduled Caste Population _2011_Total"]
    - df["Percentage of Scheduled Caste Population _2001_Total"]
)

df["Rural_Urban_Gap_2011"] = (
    df["Percentage of Scheduled Caste Population _2011_Rural"]
    - df["Percentage of Scheduled Caste Population _2011_Urban"]
)

df["Estimated_2021_Total_SC_Percentage"] = (
    df["Percentage of Scheduled Caste Population _2011_Total"]
    + df["Total_Growth"]
)

# =========================
# FEATURES AND TARGET
# =========================

feature_cols = [
    "Percentage of Scheduled Caste Population _2001_Rural",
    "Percentage of Scheduled Caste Population _2001_Urban",
    "Percentage of Scheduled Caste Population _2001_Total",
    "Percentage of Scheduled Caste Population _2011_Rural",
    "Percentage of Scheduled Caste Population _2011_Urban",
    "Percentage of Scheduled Caste Population _2011_Total",
    "Rural_Growth",
    "Urban_Growth",
    "Total_Growth",
    "Rural_Urban_Gap_2011",
]

target_col = "Estimated_2021_Total_SC_Percentage"

X = df[feature_cols]
y = df[target_col]

# =========================
# TRAIN TEST SPLIT
# =========================

X_train, X_test, y_train, y_test, district_train, district_test = train_test_split(
    X,
    y,
    df[district_col],
    test_size=0.2,
    random_state=42
)

# =========================
# SCALING
# =========================

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

scaler_path = os.path.join(OUTPUT_DIR, PREFIX + "scaler.pkl")
with open(scaler_path, "wb") as f:
    pickle.dump(scaler, f)

# =========================
# TRAIN ML MODELS
# =========================

models = {
    "RandomForest": RandomForestRegressor(
        n_estimators=200,
        random_state=42
    ),
    "GradientBoosting": GradientBoostingRegressor(
        n_estimators=200,
        learning_rate=0.05,
        random_state=42
    )
}

results = []
all_predictions = {}

best_model = None
best_model_name = None
best_r2 = -999

for name, model in models.items():
    model.fit(X_train_scaled, y_train)

    preds = model.predict(X_test_scaled)
    all_predictions[name] = preds

    mae = mean_absolute_error(y_test, preds)
    mse = mean_squared_error(y_test, preds)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_test, preds)

    results.append({
        "Model": name,
        "MAE": mae,
        "MSE": mse,
        "RMSE": rmse,
        "R2_Score": r2,
        "Accuracy_Percentage": r2 * 100
    })

    model_path = os.path.join(OUTPUT_DIR, PREFIX + name.lower() + "_model.pkl")
    with open(model_path, "wb") as f:
        pickle.dump(model, f)

    if r2 > best_r2:
        best_r2 = r2
        best_model = model
        best_model_name = name

# =========================
# DEEP LEARNING MODEL
# =========================

dl_model = Sequential([
    Dense(64, activation="relu", input_shape=(X_train_scaled.shape[1],)),
    Dropout(0.2),
    Dense(32, activation="relu"),
    Dense(16, activation="relu"),
    Dense(1)
])

dl_model.compile(
    optimizer="adam",
    loss="mse",
    metrics=["mae"]
)

early_stop = EarlyStopping(
    monitor="val_loss",
    patience=20,
    restore_best_weights=True
)

history = dl_model.fit(
    X_train_scaled,
    y_train,
    validation_split=0.2,
    epochs=300,
    batch_size=8,
    callbacks=[early_stop],
    verbose=1
)

dl_preds = dl_model.predict(X_test_scaled).flatten()
all_predictions["DeepNeuralNetwork"] = dl_preds

dl_mae = mean_absolute_error(y_test, dl_preds)
dl_mse = mean_squared_error(y_test, dl_preds)
dl_rmse = np.sqrt(dl_mse)
dl_r2 = r2_score(y_test, dl_preds)

results.append({
    "Model": "DeepNeuralNetwork",
    "MAE": dl_mae,
    "MSE": dl_mse,
    "RMSE": dl_rmse,
    "R2_Score": dl_r2,
    "Accuracy_Percentage": dl_r2 * 100
})

h5_path = os.path.join(OUTPUT_DIR, PREFIX + "deep_learning_model.h5")
dl_model.save(h5_path)

if dl_r2 > best_r2:
    best_r2 = dl_r2
    best_model_name = "DeepNeuralNetwork"

# =========================
# SAVE BEST MODEL
# =========================

best_model_path = os.path.join(OUTPUT_DIR, PREFIX + "best_model.pkl")

if best_model is not None:
    with open(best_model_path, "wb") as f:
        pickle.dump(best_model, f)

# =========================
# RESULT CSV
# =========================

results_df = pd.DataFrame(results)

results_csv_path = os.path.join(OUTPUT_DIR, PREFIX + "results.csv")
results_df.to_csv(results_csv_path, index=False)

# =========================
# PREDICTION CSV
# =========================

prediction_df = pd.DataFrame({
    "District": district_test.values,
    "Actual_2021_Estimated_SC_Percentage": y_test.values
})

for model_name, preds in all_predictions.items():
    prediction_df[f"{model_name}_Prediction"] = preds

prediction_csv_path = os.path.join(OUTPUT_DIR, PREFIX + "predictions.csv")
prediction_df.to_csv(prediction_csv_path, index=False)

# =========================
# JSON FILE
# =========================

json_data = {
    "project_name": "Scheduled Caste Population Growth & Demographic Forecasting System",
    "author": "Sagnik Patra",
    "dataset_path": DATASET_PATH,
    "output_directory": OUTPUT_DIR,
    "target_column": target_col,
    "features_used": feature_cols,
    "best_model": best_model_name,
    "best_r2_score": float(best_r2),
    "model_results": results_df.to_dict(orient="records")
}

json_path = os.path.join(OUTPUT_DIR, PREFIX + "results.json")
with open(json_path, "w") as f:
    json.dump(json_data, f, indent=4)

# =========================
# YAML FILE
# =========================

yaml_config = {
    "project": {
        "name": "Scheduled Caste Population Growth & Demographic Forecasting System",
        "author": "Sagnik Patra",
        "prefix": PREFIX
    },
    "paths": {
        "dataset_path": DATASET_PATH,
        "output_directory": OUTPUT_DIR,
        "h5_model": h5_path,
        "best_model_pkl": best_model_path,
        "scaler_pkl": scaler_path,
        "results_csv": results_csv_path,
        "predictions_csv": prediction_csv_path,
        "json_results": json_path
    },
    "training": {
        "test_size": 0.2,
        "random_state": 42,
        "target_column": target_col,
        "features": feature_cols
    },
    "best_model": best_model_name
}

yaml_path = os.path.join(OUTPUT_DIR, PREFIX + "config.yaml")
with open(yaml_path, "w") as f:
    yaml.dump(yaml_config, f, default_flow_style=False)

# =========================
# GRAPH 1: ACCURACY GRAPH
# =========================

plt.figure(figsize=(10, 6))
plt.bar(results_df["Model"], results_df["Accuracy_Percentage"])
plt.xlabel("Models")
plt.ylabel("Accuracy Percentage")
plt.title("Model Accuracy Comparison")
plt.xticks(rotation=30)
plt.tight_layout()

accuracy_graph_path = os.path.join(OUTPUT_DIR, PREFIX + "accuracy_graph.png")
plt.savefig(accuracy_graph_path)
plt.show()

# =========================
# GRAPH 2: COMPARISON GRAPH
# =========================

plt.figure(figsize=(10, 6))
plt.plot(results_df["Model"], results_df["MAE"], marker="o", label="MAE")
plt.plot(results_df["Model"], results_df["RMSE"], marker="o", label="RMSE")
plt.xlabel("Models")
plt.ylabel("Error Value")
plt.title("Model Error Comparison")
plt.legend()
plt.xticks(rotation=30)
plt.tight_layout()

comparison_graph_path = os.path.join(OUTPUT_DIR, PREFIX + "comparison_graph.png")
plt.savefig(comparison_graph_path)
plt.show()

# =========================
# GRAPH 3: HEATMAP
# =========================

corr = df[feature_cols + [target_col]].corr()

plt.figure(figsize=(12, 8))
plt.imshow(corr, interpolation="nearest")
plt.colorbar()
plt.xticks(range(len(corr.columns)), corr.columns, rotation=90)
plt.yticks(range(len(corr.columns)), corr.columns)
plt.title("Feature Correlation Heatmap")
plt.tight_layout()

heatmap_path = os.path.join(OUTPUT_DIR, PREFIX + "heatmap.png")
plt.savefig(heatmap_path)
plt.show()

# =========================
# GRAPH 4: RESULT GRAPH
# =========================

plt.figure(figsize=(10, 6))
plt.plot(results_df["Model"], results_df["R2_Score"], marker="o")
plt.xlabel("Models")
plt.ylabel("R2 Score")
plt.title("Result Graph - R2 Score by Model")
plt.xticks(rotation=30)
plt.tight_layout()

result_graph_path = os.path.join(OUTPUT_DIR, PREFIX + "result_graph.png")
plt.savefig(result_graph_path)
plt.show()

# =========================
# GRAPH 5: PREDICTION GRAPH
# =========================

best_prediction_col = f"{best_model_name}_Prediction"

plt.figure(figsize=(12, 6))
plt.plot(
    prediction_df["District"],
    prediction_df["Actual_2021_Estimated_SC_Percentage"],
    marker="o",
    label="Actual"
)

plt.plot(
    prediction_df["District"],
    prediction_df[best_prediction_col],
    marker="o",
    label="Predicted"
)

plt.xlabel("District")
plt.ylabel("SC Population Percentage")
plt.title(f"Prediction Graph - Actual vs Predicted using {best_model_name}")
plt.legend()
plt.xticks(rotation=90)
plt.tight_layout()

prediction_graph_path = os.path.join(OUTPUT_DIR, PREFIX + "prediction_graph.png")
plt.savefig(prediction_graph_path)
plt.show()

# =========================
# GRAPH 6: DEEP LEARNING LOSS GRAPH
# =========================

plt.figure(figsize=(10, 6))
plt.plot(history.history["loss"], label="Training Loss")
plt.plot(history.history["val_loss"], label="Validation Loss")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.title("Deep Learning Training Loss Graph")
plt.legend()
plt.tight_layout()

dl_loss_graph_path = os.path.join(OUTPUT_DIR, PREFIX + "dl_loss_graph.png")
plt.savefig(dl_loss_graph_path)
plt.show()

# =========================
# FINAL OUTPUT
# =========================

print("\nAll files saved successfully!")

print("\nSaved CSV Files:")
print(results_csv_path)
print(prediction_csv_path)

print("\nSaved Model Files:")
print(h5_path)
print(best_model_path)
print(scaler_path)

print("\nSaved Config Files:")
print(json_path)
print(yaml_path)

print("\nSaved Graph Files:")
print(accuracy_graph_path)
print(comparison_graph_path)
print(heatmap_path)
print(result_graph_path)
print(prediction_graph_path)
print(dl_loss_graph_path)

print("\nModel Results:")
print(results_df)
```
