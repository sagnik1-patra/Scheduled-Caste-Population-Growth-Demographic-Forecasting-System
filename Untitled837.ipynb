```python
# Scheduled Caste Population Growth & Demographic Forecasting System
# GA Feature Selection Version
# Author: Sagnik Patra

import os
import json
import yaml
import pickle
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

warnings.filterwarnings("ignore")

DATASET_PATH = r"C:\Users\NXTWAVE\Downloads\Scheduled Caste Population Growth & Demographic Forecasting System\Percentage_of_Scheduled_Caste_Population_to_Total_Population.csv"
OUTPUT_DIR = r"C:\Users\NXTWAVE\Downloads\Scheduled Caste Population Growth & Demographic Forecasting System"

os.makedirs(OUTPUT_DIR, exist_ok=True)

PREFIX = "ga_"

df = pd.read_csv(DATASET_PATH)
df.columns = df.columns.str.strip()

district_col = "District"

df["Rural_Growth"] = df["Percentage of Scheduled Caste Population _2011_Rural"] - df["Percentage of Scheduled Caste Population _2001_Rural"]
df["Urban_Growth"] = df["Percentage of Scheduled Caste Population _2011_Urban"] - df["Percentage of Scheduled Caste Population _2001_Urban"]
df["Total_Growth"] = df["Percentage of Scheduled Caste Population _2011_Total"] - df["Percentage of Scheduled Caste Population _2001_Total"]
df["Rural_Urban_Gap_2011"] = df["Percentage of Scheduled Caste Population _2011_Rural"] - df["Percentage of Scheduled Caste Population _2011_Urban"]

df["Estimated_2021_Total_SC_Percentage"] = (
    df["Percentage of Scheduled Caste Population _2011_Total"] + df["Total_Growth"]
)

feature_cols = [
    "Percentage of Scheduled Caste Population _2001_Rural",
    "Percentage of Scheduled Caste Population _2001_Urban",
    "Percentage of Scheduled Caste Population _2001_Total",
    "Percentage of Scheduled Caste Population _2011_Rural",
    "Percentage of Scheduled Caste Population _2011_Urban",
    "Percentage of Scheduled Caste Population _2011_Total",
    "Rural_Growth",
    "Urban_Growth",
    "Total_Growth",
    "Rural_Urban_Gap_2011",
]

target_col = "Estimated_2021_Total_SC_Percentage"

X = df[feature_cols]
y = df[target_col]

X_train, X_test, y_train, y_test, district_train, district_test = train_test_split(
    X, y, df[district_col], test_size=0.2, random_state=42
)

scaler = StandardScaler()
X_train_scaled_full = scaler.fit_transform(X_train)
X_test_scaled_full = scaler.transform(X_test)

scaler_path = os.path.join(OUTPUT_DIR, PREFIX + "scaler.pkl")

with open(scaler_path, "wb") as f:
    pickle.dump(scaler, f)

# =========================
# GA FEATURE SELECTION
# =========================

def fitness_function(mask, X_train_data, X_test_data, y_train_data, y_test_data):
    if np.sum(mask) == 0:
        return -9999

    selected_train = X_train_data[:, mask == 1]
    selected_test = X_test_data[:, mask == 1]

    model = RandomForestRegressor(n_estimators=100, random_state=42)
    model.fit(selected_train, y_train_data)

    preds = model.predict(selected_test)
    r2 = r2_score(y_test_data, preds)

    penalty = 0.01 * (np.sum(mask) / len(mask))
    return r2 - penalty


def tournament_selection(population, fitness_scores, tournament_size=3):
    selected_indices = np.random.choice(len(population), tournament_size, replace=False)
    best_index = selected_indices[np.argmax(fitness_scores[selected_indices])]
    return population[best_index].copy()


def crossover(parent1, parent2, crossover_rate=0.8):
    if np.random.rand() > crossover_rate:
        return parent1.copy(), parent2.copy()

    point = np.random.randint(1, len(parent1) - 1)

    child1 = np.concatenate([parent1[:point], parent2[point:]])
    child2 = np.concatenate([parent2[:point], parent1[point:]])

    return child1, child2


def mutate(individual, mutation_rate=0.1):
    for i in range(len(individual)):
        if np.random.rand() < mutation_rate:
            individual[i] = 1 - individual[i]

    if np.sum(individual) == 0:
        individual[np.random.randint(0, len(individual))] = 1

    return individual


def ga_feature_selection(
    X_train_data,
    X_test_data,
    y_train_data,
    y_test_data,
    num_features,
    population_size=30,
    generations=30,
    crossover_rate=0.8,
    mutation_rate=0.1,
    elitism_count=2
):
    population = np.random.randint(0, 2, size=(population_size, num_features))

    for individual in population:
        if np.sum(individual) == 0:
            individual[np.random.randint(0, num_features)] = 1

    best_solution = None
    best_fitness = -9999
    fitness_history = []

    for generation in range(generations):
        fitness_scores = np.array([
            fitness_function(individual, X_train_data, X_test_data, y_train_data, y_test_data)
            for individual in population
        ])

        sorted_indices = np.argsort(fitness_scores)[::-1]
        population = population[sorted_indices]
        fitness_scores = fitness_scores[sorted_indices]

        if fitness_scores[0] > best_fitness:
            best_fitness = fitness_scores[0]
            best_solution = population[0].copy()

        fitness_history.append(best_fitness)

        new_population = []

        elites = population[:elitism_count]
        for elite in elites:
            new_population.append(elite.copy())

        while len(new_population) < population_size:
            parent1 = tournament_selection(population, fitness_scores)
            parent2 = tournament_selection(population, fitness_scores)

            child1, child2 = crossover(parent1, parent2, crossover_rate)

            child1 = mutate(child1, mutation_rate)
            child2 = mutate(child2, mutation_rate)

            new_population.append(child1)

            if len(new_population) < population_size:
                new_population.append(child2)

        population = np.array(new_population)

        print(
            f"Generation {generation + 1}/{generations} | "
            f"Best GA Fitness: {best_fitness:.4f}"
        )

    return best_solution, best_fitness, fitness_history


selected_mask, best_ga_fitness, ga_history = ga_feature_selection(
    X_train_scaled_full,
    X_test_scaled_full,
    y_train,
    y_test,
    num_features=X_train_scaled_full.shape[1]
)

selected_features = [
    feature for feature, selected in zip(feature_cols, selected_mask)
    if selected == 1
]

X_train_scaled = X_train_scaled_full[:, selected_mask == 1]
X_test_scaled = X_test_scaled_full[:, selected_mask == 1]

# =========================
# TRAIN MODELS
# =========================

models = {
    "GA_RandomForest": RandomForestRegressor(n_estimators=200, random_state=42),
    "GA_GradientBoosting": GradientBoostingRegressor(
        n_estimators=200,
        learning_rate=0.05,
        random_state=42
    )
}

results = []
all_predictions = {}

best_model = None
best_model_name = None
best_r2 = -999

for name, model in models.items():
    model.fit(X_train_scaled, y_train)

    preds = model.predict(X_test_scaled)
    all_predictions[name] = preds

    mae = mean_absolute_error(y_test, preds)
    mse = mean_squared_error(y_test, preds)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_test, preds)

    results.append({
        "Model": name,
        "MAE": mae,
        "MSE": mse,
        "RMSE": rmse,
        "R2_Score": r2,
        "Accuracy_Percentage": r2 * 100
    })

    model_path = os.path.join(OUTPUT_DIR, PREFIX + name.lower() + "_model.pkl")

    with open(model_path, "wb") as f:
        pickle.dump(model, f)

    if r2 > best_r2:
        best_r2 = r2
        best_model = model
        best_model_name = name

# =========================
# DEEP LEARNING MODEL
# =========================

dl_model = Sequential([
    Dense(64, activation="relu", input_shape=(X_train_scaled.shape[1],)),
    Dropout(0.2),
    Dense(32, activation="relu"),
    Dense(16, activation="relu"),
    Dense(1)
])

dl_model.compile(optimizer="adam", loss="mse", metrics=["mae"])

early_stop = EarlyStopping(
    monitor="val_loss",
    patience=20,
    restore_best_weights=True
)

history = dl_model.fit(
    X_train_scaled,
    y_train,
    validation_split=0.2,
    epochs=300,
    batch_size=8,
    callbacks=[early_stop],
    verbose=1
)

dl_preds = dl_model.predict(X_test_scaled).flatten()
all_predictions["GA_DeepNeuralNetwork"] = dl_preds

dl_mae = mean_absolute_error(y_test, dl_preds)
dl_mse = mean_squared_error(y_test, dl_preds)
dl_rmse = np.sqrt(dl_mse)
dl_r2 = r2_score(y_test, dl_preds)

results.append({
    "Model": "GA_DeepNeuralNetwork",
    "MAE": dl_mae,
    "MSE": dl_mse,
    "RMSE": dl_rmse,
    "R2_Score": dl_r2,
    "Accuracy_Percentage": dl_r2 * 100
})

h5_path = os.path.join(OUTPUT_DIR, PREFIX + "deep_learning_model.h5")
dl_model.save(h5_path)

if dl_r2 > best_r2:
    best_r2 = dl_r2
    best_model_name = "GA_DeepNeuralNetwork"

best_model_path = os.path.join(OUTPUT_DIR, PREFIX + "best_model.pkl")

if best_model is not None:
    with open(best_model_path, "wb") as f:
        pickle.dump(best_model, f)

# =========================
# SAVE FILES
# =========================

feature_selection_data = {
    "algorithm": "Genetic Algorithm",
    "selected_mask": selected_mask.tolist(),
    "selected_features": selected_features,
    "best_ga_fitness": float(best_ga_fitness),
    "ga_history": [float(x) for x in ga_history]
}

feature_selection_path = os.path.join(OUTPUT_DIR, PREFIX + "feature_selection.json")

with open(feature_selection_path, "w") as f:
    json.dump(feature_selection_data, f, indent=4)

results_df = pd.DataFrame(results)

results_csv_path = os.path.join(OUTPUT_DIR, PREFIX + "results.csv")
results_df.to_csv(results_csv_path, index=False)

prediction_df = pd.DataFrame({
    "District": district_test.values,
    "Actual_2021_Estimated_SC_Percentage": y_test.values
})

for model_name, preds in all_predictions.items():
    prediction_df[f"{model_name}_Prediction"] = preds

prediction_csv_path = os.path.join(OUTPUT_DIR, PREFIX + "predictions.csv")
prediction_df.to_csv(prediction_csv_path, index=False)

json_data = {
    "project_name": "Scheduled Caste Population Growth & Demographic Forecasting System",
    "author": "Sagnik Patra",
    "algorithm": "Genetic Algorithm Feature Selection",
    "dataset_path": DATASET_PATH,
    "output_directory": OUTPUT_DIR,
    "target_column": target_col,
    "all_features": feature_cols,
    "selected_features": selected_features,
    "best_model": best_model_name,
    "best_r2_score": float(best_r2),
    "best_ga_fitness": float(best_ga_fitness),
    "model_results": results_df.to_dict(orient="records")
}

json_path = os.path.join(OUTPUT_DIR, PREFIX + "results.json")

with open(json_path, "w") as f:
    json.dump(json_data, f, indent=4)

yaml_config = {
    "project": {
        "name": "Scheduled Caste Population Growth & Demographic Forecasting System",
        "author": "Sagnik Patra",
        "prefix": PREFIX,
        "algorithm": "GA"
    },
    "paths": {
        "dataset_path": DATASET_PATH,
        "output_directory": OUTPUT_DIR,
        "h5_model": h5_path,
        "best_model_pkl": best_model_path,
        "scaler_pkl": scaler_path,
        "results_csv": results_csv_path,
        "predictions_csv": prediction_csv_path,
        "json_results": json_path,
        "feature_selection_json": feature_selection_path
    },
    "training": {
        "test_size": 0.2,
        "random_state": 42,
        "target_column": target_col,
        "all_features": feature_cols,
        "selected_features": selected_features
    },
    "ga_parameters": {
        "population_size": 30,
        "generations": 30,
        "crossover_rate": 0.8,
        "mutation_rate": 0.1,
        "elitism_count": 2
    },
    "best_model": best_model_name
}

yaml_path = os.path.join(OUTPUT_DIR, PREFIX + "config.yaml")

with open(yaml_path, "w") as f:
    yaml.dump(yaml_config, f, default_flow_style=False)

# =========================
# GRAPHS
# =========================

plt.figure(figsize=(10, 6))
plt.bar(results_df["Model"], results_df["Accuracy_Percentage"])
plt.xlabel("Models")
plt.ylabel("Accuracy Percentage")
plt.title("GA Model Accuracy Graph")
plt.xticks(rotation=30)
plt.tight_layout()
accuracy_graph_path = os.path.join(OUTPUT_DIR, PREFIX + "accuracy_graph.png")
plt.savefig(accuracy_graph_path)
plt.show()

plt.figure(figsize=(10, 6))
plt.plot(results_df["Model"], results_df["MAE"], marker="o", label="MAE")
plt.plot(results_df["Model"], results_df["RMSE"], marker="o", label="RMSE")
plt.xlabel("Models")
plt.ylabel("Error Value")
plt.title("GA Comparison Graph")
plt.legend()
plt.xticks(rotation=30)
plt.tight_layout()
comparison_graph_path = os.path.join(OUTPUT_DIR, PREFIX + "comparison_graph.png")
plt.savefig(comparison_graph_path)
plt.show()

corr = df[selected_features + [target_col]].corr()

plt.figure(figsize=(12, 8))
plt.imshow(corr, interpolation="nearest")
plt.colorbar()
plt.xticks(range(len(corr.columns)), corr.columns, rotation=90)
plt.yticks(range(len(corr.columns)), corr.columns)
plt.title("GA Selected Feature Heatmap")
plt.tight_layout()
heatmap_path = os.path.join(OUTPUT_DIR, PREFIX + "heatmap.png")
plt.savefig(heatmap_path)
plt.show()

plt.figure(figsize=(10, 6))
plt.plot(results_df["Model"], results_df["R2_Score"], marker="o")
plt.xlabel("Models")
plt.ylabel("R2 Score")
plt.title("GA Result Graph")
plt.xticks(rotation=30)
plt.tight_layout()
result_graph_path = os.path.join(OUTPUT_DIR, PREFIX + "result_graph.png")
plt.savefig(result_graph_path)
plt.show()

best_prediction_col = f"{best_model_name}_Prediction"

plt.figure(figsize=(12, 6))
plt.plot(
    prediction_df["District"],
    prediction_df["Actual_2021_Estimated_SC_Percentage"],
    marker="o",
    label="Actual"
)
plt.plot(
    prediction_df["District"],
    prediction_df[best_prediction_col],
    marker="o",
    label="Predicted"
)
plt.xlabel("District")
plt.ylabel("SC Population Percentage")
plt.title(f"GA Prediction Graph - {best_model_name}")
plt.legend()
plt.xticks(rotation=90)
plt.tight_layout()
prediction_graph_path = os.path.join(OUTPUT_DIR, PREFIX + "prediction_graph.png")
plt.savefig(prediction_graph_path)
plt.show()

plt.figure(figsize=(10, 6))
plt.plot(ga_history, marker="o")
plt.xlabel("Generation")
plt.ylabel("Best Fitness")
plt.title("GA Optimization Progress Graph")
plt.tight_layout()
optimization_graph_path = os.path.join(OUTPUT_DIR, PREFIX + "optimization_graph.png")
plt.savefig(optimization_graph_path)
plt.show()

plt.figure(figsize=(10, 6))
plt.plot(history.history["loss"], label="Training Loss")
plt.plot(history.history["val_loss"], label="Validation Loss")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.title("GA Deep Learning Loss Graph")
plt.legend()
plt.tight_layout()
dl_loss_graph_path = os.path.join(OUTPUT_DIR, PREFIX + "dl_loss_graph.png")
plt.savefig(dl_loss_graph_path)
plt.show()

print("\nGA execution completed successfully!")
print("\nSaved prefix:", PREFIX)
print("\nSelected Features:", selected_features)
print("\nResults CSV:", results_csv_path)
print("Predictions CSV:", prediction_csv_path)
print("H5 Model:", h5_path)
print("Best PKL Model:", best_model_path)
print("YAML:", yaml_path)
print("JSON:", json_path)
print("\nModel Results:")
print(results_df)
```
