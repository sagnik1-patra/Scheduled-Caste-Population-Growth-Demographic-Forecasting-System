```python
# Scheduled Caste Population Growth & Demographic Forecasting System
# CSA Feature Selection Version
# Author: Sagnik Patra

import os
import json
import yaml
import pickle
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

warnings.filterwarnings("ignore")

# =========================
# PATHS
# =========================

DATASET_PATH = r"C:\Users\NXTWAVE\Downloads\Scheduled Caste Population Growth & Demographic Forecasting System\Percentage_of_Scheduled_Caste_Population_to_Total_Population.csv"

OUTPUT_DIR = r"C:\Users\NXTWAVE\Downloads\Scheduled Caste Population Growth & Demographic Forecasting System"

os.makedirs(OUTPUT_DIR, exist_ok=True)

PREFIX = "csa_"

# =========================
# LOAD DATA
# =========================

df = pd.read_csv(DATASET_PATH)
df.columns = df.columns.str.strip()

district_col = "District"

# =========================
# FEATURE ENGINEERING
# =========================

df["Rural_Growth"] = (
    df["Percentage of Scheduled Caste Population _2011_Rural"]
    - df["Percentage of Scheduled Caste Population _2001_Rural"]
)

df["Urban_Growth"] = (
    df["Percentage of Scheduled Caste Population _2011_Urban"]
    - df["Percentage of Scheduled Caste Population _2001_Urban"]
)

df["Total_Growth"] = (
    df["Percentage of Scheduled Caste Population _2011_Total"]
    - df["Percentage of Scheduled Caste Population _2001_Total"]
)

df["Rural_Urban_Gap_2011"] = (
    df["Percentage of Scheduled Caste Population _2011_Rural"]
    - df["Percentage of Scheduled Caste Population _2011_Urban"]
)

df["Estimated_2021_Total_SC_Percentage"] = (
    df["Percentage of Scheduled Caste Population _2011_Total"]
    + df["Total_Growth"]
)

feature_cols = [
    "Percentage of Scheduled Caste Population _2001_Rural",
    "Percentage of Scheduled Caste Population _2001_Urban",
    "Percentage of Scheduled Caste Population _2001_Total",
    "Percentage of Scheduled Caste Population _2011_Rural",
    "Percentage of Scheduled Caste Population _2011_Urban",
    "Percentage of Scheduled Caste Population _2011_Total",
    "Rural_Growth",
    "Urban_Growth",
    "Total_Growth",
    "Rural_Urban_Gap_2011",
]

target_col = "Estimated_2021_Total_SC_Percentage"

X = df[feature_cols]
y = df[target_col]

# =========================
# TRAIN TEST SPLIT
# =========================

X_train, X_test, y_train, y_test, district_train, district_test = train_test_split(
    X,
    y,
    df[district_col],
    test_size=0.2,
    random_state=42
)

# =========================
# SCALING
# =========================

scaler = StandardScaler()
X_train_scaled_full = scaler.fit_transform(X_train)
X_test_scaled_full = scaler.transform(X_test)

scaler_path = os.path.join(OUTPUT_DIR, PREFIX + "scaler.pkl")

with open(scaler_path, "wb") as f:
    pickle.dump(scaler, f)

# =========================
# CSA FEATURE SELECTION
# =========================

def fitness_function(mask, X_train_data, X_test_data, y_train_data, y_test_data):
    if np.sum(mask) == 0:
        return -9999

    selected_train = X_train_data[:, mask == 1]
    selected_test = X_test_data[:, mask == 1]

    model = RandomForestRegressor(
        n_estimators=100,
        random_state=42
    )

    model.fit(selected_train, y_train_data)
    preds = model.predict(selected_test)

    r2 = r2_score(y_test_data, preds)

    feature_penalty = 0.01 * (np.sum(mask) / len(mask))

    return r2 - feature_penalty


def csa_feature_selection(
    X_train_data,
    X_test_data,
    y_train_data,
    y_test_data,
    num_features,
    population_size=20,
    generations=30,
    clone_factor=5,
    mutation_rate=0.2,
    replacement_rate=0.25
):
    population = np.random.randint(0, 2, size=(population_size, num_features))

    for individual in population:
        if np.sum(individual) == 0:
            individual[np.random.randint(0, num_features)] = 1

    best_solution = None
    best_fitness = -9999
    fitness_history = []

    for generation in range(generations):
        fitness_scores = np.array([
            fitness_function(
                individual,
                X_train_data,
                X_test_data,
                y_train_data,
                y_test_data
            )
            for individual in population
        ])

        sorted_indices = np.argsort(fitness_scores)[::-1]
        population = population[sorted_indices]
        fitness_scores = fitness_scores[sorted_indices]

        if fitness_scores[0] > best_fitness:
            best_fitness = fitness_scores[0]
            best_solution = population[0].copy()

        fitness_history.append(best_fitness)

        elite_count = max(2, population_size // 4)
        selected_antibodies = population[:elite_count]

        clone_pool = []

        for rank, antibody in enumerate(selected_antibodies):
            number_of_clones = clone_factor * (elite_count - rank)

            for _ in range(number_of_clones):
                clone = antibody.copy()

                adaptive_mutation_rate = mutation_rate * (rank + 1)

                for i in range(num_features):
                    if np.random.rand() < adaptive_mutation_rate:
                        clone[i] = 1 - clone[i]

                if np.sum(clone) == 0:
                    clone[np.random.randint(0, num_features)] = 1

                clone_pool.append(clone)

        clone_pool = np.array(clone_pool)

        clone_fitness = np.array([
            fitness_function(
                clone,
                X_train_data,
                X_test_data,
                y_train_data,
                y_test_data
            )
            for clone in clone_pool
        ])

        best_clone_indices = np.argsort(clone_fitness)[::-1]

        improved_population = clone_pool[best_clone_indices[:population_size]]

        replace_count = int(population_size * replacement_rate)

        random_new = np.random.randint(
            0,
            2,
            size=(replace_count, num_features)
        )

        for individual in random_new:
            if np.sum(individual) == 0:
                individual[np.random.randint(0, num_features)] = 1

        population = improved_population.copy()

        population[-replace_count:] = random_new

        print(
            f"Generation {generation + 1}/{generations} | "
            f"Best CSA Fitness: {best_fitness:.4f}"
        )

    return best_solution, best_fitness, fitness_history


selected_mask, best_csa_fitness, csa_history = csa_feature_selection(
    X_train_scaled_full,
    X_test_scaled_full,
    y_train,
    y_test,
    num_features=X_train_scaled_full.shape[1]
)

selected_features = [
    feature for feature, selected in zip(feature_cols, selected_mask)
    if selected == 1
]

X_train_scaled = X_train_scaled_full[:, selected_mask == 1]
X_test_scaled = X_test_scaled_full[:, selected_mask == 1]

# =========================
# TRAIN MODELS ON CSA FEATURES
# =========================

models = {
    "CSA_RandomForest": RandomForestRegressor(
        n_estimators=200,
        random_state=42
    ),
    "CSA_GradientBoosting": GradientBoostingRegressor(
        n_estimators=200,
        learning_rate=0.05,
        random_state=42
    )
}

results = []
all_predictions = {}

best_model = None
best_model_name = None
best_r2 = -999

for name, model in models.items():
    model.fit(X_train_scaled, y_train)

    preds = model.predict(X_test_scaled)
    all_predictions[name] = preds

    mae = mean_absolute_error(y_test, preds)
    mse = mean_squared_error(y_test, preds)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_test, preds)

    results.append({
        "Model": name,
        "MAE": mae,
        "MSE": mse,
        "RMSE": rmse,
        "R2_Score": r2,
        "Accuracy_Percentage": r2 * 100
    })

    model_path = os.path.join(
        OUTPUT_DIR,
        PREFIX + name.lower() + "_model.pkl"
    )

    with open(model_path, "wb") as f:
        pickle.dump(model, f)

    if r2 > best_r2:
        best_r2 = r2
        best_model = model
        best_model_name = name

# =========================
# DEEP LEARNING MODEL
# =========================

dl_model = Sequential([
    Dense(64, activation="relu", input_shape=(X_train_scaled.shape[1],)),
    Dropout(0.2),
    Dense(32, activation="relu"),
    Dense(16, activation="relu"),
    Dense(1)
])

dl_model.compile(
    optimizer="adam",
    loss="mse",
    metrics=["mae"]
)

early_stop = EarlyStopping(
    monitor="val_loss",
    patience=20,
    restore_best_weights=True
)

history = dl_model.fit(
    X_train_scaled,
    y_train,
    validation_split=0.2,
    epochs=300,
    batch_size=8,
    callbacks=[early_stop],
    verbose=1
)

dl_preds = dl_model.predict(X_test_scaled).flatten()

all_predictions["CSA_DeepNeuralNetwork"] = dl_preds

dl_mae = mean_absolute_error(y_test, dl_preds)
dl_mse = mean_squared_error(y_test, dl_preds)
dl_rmse = np.sqrt(dl_mse)
dl_r2 = r2_score(y_test, dl_preds)

results.append({
    "Model": "CSA_DeepNeuralNetwork",
    "MAE": dl_mae,
    "MSE": dl_mse,
    "RMSE": dl_rmse,
    "R2_Score": dl_r2,
    "Accuracy_Percentage": dl_r2 * 100
})

h5_path = os.path.join(OUTPUT_DIR, PREFIX + "deep_learning_model.h5")
dl_model.save(h5_path)

if dl_r2 > best_r2:
    best_r2 = dl_r2
    best_model_name = "CSA_DeepNeuralNetwork"

# =========================
# SAVE BEST MODEL
# =========================

best_model_path = os.path.join(OUTPUT_DIR, PREFIX + "best_model.pkl")

if best_model is not None:
    with open(best_model_path, "wb") as f:
        pickle.dump(best_model, f)

# =========================
# SAVE CSA FEATURE SELECTION JSON
# =========================

feature_selection_data = {
    "algorithm": "Clonal Selection Algorithm",
    "selected_mask": selected_mask.tolist(),
    "selected_features": selected_features,
    "best_csa_fitness": float(best_csa_fitness),
    "csa_history": [float(x) for x in csa_history]
}

feature_selection_path = os.path.join(
    OUTPUT_DIR,
    PREFIX + "feature_selection.json"
)

with open(feature_selection_path, "w") as f:
    json.dump(feature_selection_data, f, indent=4)

# =========================
# SAVE CSV FILES
# =========================

results_df = pd.DataFrame(results)

results_csv_path = os.path.join(OUTPUT_DIR, PREFIX + "results.csv")
results_df.to_csv(results_csv_path, index=False)

prediction_df = pd.DataFrame({
    "District": district_test.values,
    "Actual_2021_Estimated_SC_Percentage": y_test.values
})

for model_name, preds in all_predictions.items():
    prediction_df[f"{model_name}_Prediction"] = preds

prediction_csv_path = os.path.join(OUTPUT_DIR, PREFIX + "predictions.csv")
prediction_df.to_csv(prediction_csv_path, index=False)

# =========================
# SAVE JSON RESULT FILE
# =========================

json_data = {
    "project_name": "Scheduled Caste Population Growth & Demographic Forecasting System",
    "author": "Sagnik Patra",
    "algorithm": "Clonal Selection Algorithm Feature Selection",
    "dataset_path": DATASET_PATH,
    "output_directory": OUTPUT_DIR,
    "target_column": target_col,
    "all_features": feature_cols,
    "selected_features": selected_features,
    "best_model": best_model_name,
    "best_r2_score": float(best_r2),
    "best_csa_fitness": float(best_csa_fitness),
    "model_results": results_df.to_dict(orient="records")
}

json_path = os.path.join(OUTPUT_DIR, PREFIX + "results.json")

with open(json_path, "w") as f:
    json.dump(json_data, f, indent=4)

# =========================
# SAVE YAML CONFIG FILE
# =========================

yaml_config = {
    "project": {
        "name": "Scheduled Caste Population Growth & Demographic Forecasting System",
        "author": "Sagnik Patra",
        "prefix": PREFIX,
        "algorithm": "CSA"
    },
    "paths": {
        "dataset_path": DATASET_PATH,
        "output_directory": OUTPUT_DIR,
        "h5_model": h5_path,
        "best_model_pkl": best_model_path,
        "scaler_pkl": scaler_path,
        "results_csv": results_csv_path,
        "predictions_csv": prediction_csv_path,
        "json_results": json_path,
        "feature_selection_json": feature_selection_path
    },
    "training": {
        "test_size": 0.2,
        "random_state": 42,
        "target_column": target_col,
        "all_features": feature_cols,
        "selected_features": selected_features
    },
    "csa_parameters": {
        "population_size": 20,
        "generations": 30,
        "clone_factor": 5,
        "mutation_rate": 0.2,
        "replacement_rate": 0.25
    },
    "best_model": best_model_name
}

yaml_path = os.path.join(OUTPUT_DIR, PREFIX + "config.yaml")

with open(yaml_path, "w") as f:
    yaml.dump(yaml_config, f, default_flow_style=False)

# =========================
# GRAPHS
# =========================

plt.figure(figsize=(10, 6))
plt.bar(results_df["Model"], results_df["Accuracy_Percentage"])
plt.xlabel("Models")
plt.ylabel("Accuracy Percentage")
plt.title("CSA Model Accuracy Graph")
plt.xticks(rotation=30)
plt.tight_layout()

accuracy_graph_path = os.path.join(OUTPUT_DIR, PREFIX + "accuracy_graph.png")
plt.savefig(accuracy_graph_path)
plt.show()

plt.figure(figsize=(10, 6))
plt.plot(results_df["Model"], results_df["MAE"], marker="o", label="MAE")
plt.plot(results_df["Model"], results_df["RMSE"], marker="o", label="RMSE")
plt.xlabel("Models")
plt.ylabel("Error Value")
plt.title("CSA Comparison Graph")
plt.legend()
plt.xticks(rotation=30)
plt.tight_layout()

comparison_graph_path = os.path.join(OUTPUT_DIR, PREFIX + "comparison_graph.png")
plt.savefig(comparison_graph_path)
plt.show()

corr = df[selected_features + [target_col]].corr()

plt.figure(figsize=(12, 8))
plt.imshow(corr, interpolation="nearest")
plt.colorbar()
plt.xticks(range(len(corr.columns)), corr.columns, rotation=90)
plt.yticks(range(len(corr.columns)), corr.columns)
plt.title("CSA Selected Feature Heatmap")
plt.tight_layout()

heatmap_path = os.path.join(OUTPUT_DIR, PREFIX + "heatmap.png")
plt.savefig(heatmap_path)
plt.show()

plt.figure(figsize=(10, 6))
plt.plot(results_df["Model"], results_df["R2_Score"], marker="o")
plt.xlabel("Models")
plt.ylabel("R2 Score")
plt.title("CSA Result Graph")
plt.xticks(rotation=30)
plt.tight_layout()

result_graph_path = os.path.join(OUTPUT_DIR, PREFIX + "result_graph.png")
plt.savefig(result_graph_path)
plt.show()

best_prediction_col = f"{best_model_name}_Prediction"

plt.figure(figsize=(12, 6))
plt.plot(
    prediction_df["District"],
    prediction_df["Actual_2021_Estimated_SC_Percentage"],
    marker="o",
    label="Actual"
)

plt.plot(
    prediction_df["District"],
    prediction_df[best_prediction_col],
    marker="o",
    label="Predicted"
)

plt.xlabel("District")
plt.ylabel("SC Population Percentage")
plt.title(f"CSA Prediction Graph - {best_model_name}")
plt.legend()
plt.xticks(rotation=90)
plt.tight_layout()

prediction_graph_path = os.path.join(OUTPUT_DIR, PREFIX + "prediction_graph.png")
plt.savefig(prediction_graph_path)
plt.show()

plt.figure(figsize=(10, 6))
plt.plot(csa_history, marker="o")
plt.xlabel("Generation")
plt.ylabel("Best Fitness")
plt.title("CSA Optimization Progress Graph")
plt.tight_layout()

optimization_graph_path = os.path.join(OUTPUT_DIR, PREFIX + "optimization_graph.png")
plt.savefig(optimization_graph_path)
plt.show()

plt.figure(figsize=(10, 6))
plt.plot(history.history["loss"], label="Training Loss")
plt.plot(history.history["val_loss"], label="Validation Loss")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.title("CSA Deep Learning Loss Graph")
plt.legend()
plt.tight_layout()

dl_loss_graph_path = os.path.join(OUTPUT_DIR, PREFIX + "dl_loss_graph.png")
plt.savefig(dl_loss_graph_path)
plt.show()

# =========================
# FINAL OUTPUT
# =========================

print("\nCSA execution completed successfully!")

print("\nSelected Features:")
for feature in selected_features:
    print("-", feature)

print("\nSaved Files:")
print(results_csv_path)
print(prediction_csv_path)
print(h5_path)
print(best_model_path)
print(scaler_path)
print(json_path)
print(yaml_path)
print(feature_selection_path)

print("\nSaved Graphs:")
print(accuracy_graph_path)
print(comparison_graph_path)
print(heatmap_path)
print(result_graph_path)
print(prediction_graph_path)
print(optimization_graph_path)
print(dl_loss_graph_path)

print("\nModel Results:")
print(results_df)
```
